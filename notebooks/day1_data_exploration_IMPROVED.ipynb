```python
# DAY 1: DATA EXPLORATION (IMPROVED)
# UCI Heart Disease Prediction
# PURPOSE:
#   - Load and inspect the dataset
#   - Understand data structure
#   - Identify data quality issues
#   - Generate initial summary report


import pandas as pd
import numpy as np
import os
import json
from datetime import datetime

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# SECTION 1: LOAD DATA
print("LOAD DATA")
data_path = 'data/heart_disease_uci.csv'
df = pd.read_csv(data_path, na_values='?')

print(f"✓ Data loaded from: {data_path}")
print(f"  Shape: {df.shape[0]} rows × {df.shape[1]} columns")
```

```python

# SECTION 2: BASIC DATA INSPECTION
print("DATA INSPECTION")

print(f"\n Column Information:")
print(df.dtypes)

print(f"\n\n\n First 5 Rows:\n")
print(df.head())

print(f"\n\n\n Statistical Summary:\n")
print(df.describe())
```

```python
# SECTION 3: DATA QUALITY ASSESSMENT
print("DATA QUALITY ASSESSMENT")

print(f"\n Missing Values:")
missing = df.isnull().sum()
if missing.sum() > 0:
    print(missing[missing > 0])
else:
    print("  ✓ No missing values found")

print(f"\n\n\n Duplicate Rows:")
duplicates = df.duplicated().sum()
print(f"  Found: {duplicates} duplicate rows")

print(f"\n\n\n Data Types:")
print(f"  Numeric:  {len(df.select_dtypes(include=[np.number]).columns)} columns")
print(f"  Object:   {len(df.select_dtypes(exclude=[np.number]).columns)} columns")
```

```python
# SECTION 4: TARGET VARIABLE ANALYSIS
print("TARGET VARIABLE ANALYSIS")
#'num' is the target variable (heart disease)
target_col = 'num'

if target_col in df.columns:
    print(f"\n Target Variable: {target_col}")
    print(f"\n\n\n  Value Counts:")
    print(df[target_col].value_counts().sort_index())
    
    print(f"\n\n\n  Distribution (%):")
    dist = (df[target_col].value_counts(normalize=True).sort_index() * 100).round(2)
    for idx, val in dist.items():
        print(f"    {idx}: {val}%")
```

```python
# SECTION 5: FEATURE OVERVIEW
print("FEATURE OVERVIEW")

print(f"\n\n\nNumeric Features:")
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
for col in numeric_cols:
    print(f"  • {col:20s} - Range: [{df[col].min():8.2f}, {df[col].max():8.2f}]")

print(f"\n\n\nCategorical Features:")
categorical_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()
for col in categorical_cols:
    unique_count = df[col].nunique()
    print(f"  • {col:20s} - Unique values: {unique_count}")
```

```python
# SECTION 6: GENERATE EXPLORATION SUMMARY

print("GENERATING SUMMARY REPORT")

exploration_summary = {
    'exploration_date': datetime.now().isoformat(),
    'dataset': {
        'name': 'UCI Heart Disease',
        'rows': int(df.shape[0]),
        'columns': int(df.shape[1]),
        'shape': f"{df.shape[0]} × {df.shape[1]}"
    },
    'data_quality': {
        'missing_values': int(df.isnull().sum().sum()),
        'duplicate_rows': int(df.duplicated().sum()),
        'numeric_columns': len(numeric_cols),
        'categorical_columns': len(categorical_cols)
    },
    'columns': {
        col: {
            'dtype': str(df[col].dtype),
            'unique_values': int(df[col].nunique()),
            'missing': int(df[col].isnull().sum())
        }
        for col in df.columns
    },
    'target_distribution': df[target_col].value_counts().to_dict() if target_col in df.columns else {},
    'numeric_ranges': {
        col: {
            'min': float(df[col].min()),
            'max': float(df[col].max()),
            'mean': float(df[col].mean()),
            'std': float(df[col].std())
        }
        for col in numeric_cols
    }
}

# Create output directory
os.makedirs('outputs/reports', exist_ok=True)

# Save summary
summary_path = 'outputs/reports/day1_data_summary.json'
with open(summary_path, 'w') as f:
    json.dump(exploration_summary, f, indent=2)

print(f"✓ Saved: {summary_path}")
```

```python
# SECTION 7: SAVE CLEANED DATA REFERENCE

info_text = f"""
DATASET: UCI Heart Disease
Generated: {datetime.now()}

DIMENSIONS: {df.shape[0]} rows × {df.shape[1]} columns

COLUMNS:
{df.columns.tolist()}

MISSING VALUES:
{df.isnull().sum().sum()} total missing values

DATA TYPES:
{df.dtypes.to_dict()}
"""

os.makedirs('outputs', exist_ok=True)
with open('outputs/DATA_INFO.txt', 'w') as f:
    f.write(info_text)
```
